```python
X = df[df['origin']=='Asia'][['vehicle_weight','model_year']].head(7)
```

```python
X = X.reset_index(drop=True)
```

```python
X = X.to_numpy()
```

```python
X
```

```python
XTX = X.T @ X
```

```python
XTX
```

```python
XTX_inv = np.linalg.inv(XTX)
```

```python
XTX_inv
```

```python
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])
```

```python
w = XTX_inv @ X.T @ y
```

```python
result = np.sum(w)
```
